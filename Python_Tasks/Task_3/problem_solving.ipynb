# Q1. Find the Frequency of Each Word in a Sentence

## Problem
Count how many times each word appears in a sentence and store the result in a dictionary.

## Pseudocode
1. Start.
2. Take a sentence as input.
3. Split the sentence into individual words.
4. Create an empty dictionary.
5. Visit each word in the sentence.
6. If the word already exists in the dictionary, increase its count by 1.
7. Otherwise, add the word to the dictionary with a count of 1.
8. Return the dictionary.
9. Stop.

# Q10. Find Two Numbers Whose Sum Equals a Target

# Q11. Find the Number Closest to Zero

# Q13. Rotate a List to the Right by K Positions
1. Start.
2. Take a list and the number of positions K.
3. Find the length of the list.
4. If the list is empty, return the empty list.
5. Calculate K modulo the length of the list.
6. Take the last K elements of the list.
7. Take the remaining elements from the beginning.
8. Join the last K elements and the remaining elements.
9. Return the rotated list.
10. Stop.

# Q14. Find the Maximum Sum of Two Consecutive Elements

1. Start.
2. Take a list containing at least two elements.
3. Assume the sum of the first two elements is the maximum sum.
4. Traverse the list from the second element to the last element.
5. Add each element to the previous element.
6. If the current sum is greater than the maximum sum, update it.
7. Return the maximum sum.
8. Stop.

# Q15. Find All Pairs of Numbers Whose Sum Equals a Target
1. Start.
2. Take a list of numbers and a target.
3. Create an empty list called result.
4. Traverse each element using its index.
5. Compare it with every element that comes after it.
6. If the two numbers add up to the target:
   a. Arrange the two numbers in ascending order.
   b. Check whether the pair is already in the result.
   c. If it is not present, add the pair to the result.
7. Sort the result by the first element of each pair.
8. Return the result.
9. Stop.
