## Q1. Find the Shortest Word in a Sentence

## Q16. Find the Most Frequent Word in a Sentence

```python
sentence = "Python is easy and Python is powerful"

words = sentence.lower().split()

d = {}

for word in words:
    if word in d:
        d[word] += 1
    else:
        d[word] = 1

max_word = max(d, key=d.get)

print(max_word)
```

## Q20. Find the Longest Word That Appears Only Once

```python
str_7 = "Python makes coding useful Python".lower()
s = str_7.split()
```

```python
d = {}

for word in s:
    if word in d:
        d[word] += 1
    else:
        d[word] = 1

longest = ""

for word in s:
    if d[word] == 1:
        if len(word) > len(longest):
            longest = word

print(longest)
```
